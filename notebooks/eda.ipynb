# 解約データの下見

生データをざっと眺めて、どの列が効きそうかを見る。
ここで固まった処理は `src/features.py` に移すこと。

```python
import sys
from pathlib import Path

import pandas as pd

ROOT = Path.cwd().parent
sys.path.insert(0, str(ROOT))

train = pd.read_csv(ROOT / "data" / "raw" / "train.csv")
train.head()
```

```python
train["churn"].mean()
```

## 契約形態ごとの解約率

```python
train.groupby("contract")["churn"].agg(["mean", "size"])
```

## 契約期間と解約率

契約期間を10か月ごとに区切って解約率を見る。

```python
bins = range(0, 70, 10)
train.groupby(pd.cut(train["tenure"], bins))["churn"].mean()
```

## サポートへの問い合わせ回数

問い合わせが多い顧客ほど解約しやすいか。

```python
train.groupby("support_calls")["churn"].agg(["mean", "size"])
```
